# SMM StateGraph

Linear LangGraph workflow (see `reference/smm.png`):

`START → Data Extractor → Account Scraper & Prompt Modifier → Prompt 1 … Prompt 6 → END`

All logic lives in the `smm/` package; this notebook is for exploring it interactively. Keys are read from `.env`.

```python
from pathlib import Path
from smm import Settings, build_graph, run_pipeline

settings = Settings()          # LLM_PROVIDER / keys from .env
settings.validate()
```

```python
# Visualize the graph
from IPython.display import Image, display

app = build_graph()
try:
    display(Image(app.get_graph().draw_mermaid_png()))
except Exception as e:
    print("Could not render graph image:", e)
    print(app.get_graph().draw_mermaid())
```

```python
# Run the full pipeline on input.txt (costs Apify credits + LLM tokens)
final_state = run_pipeline({"input_text": Path("input.txt").read_text()}, settings)
print(final_state["output_md_path"])
```
